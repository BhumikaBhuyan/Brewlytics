```python
df["revenue"] = df["transaction_qty"] * df["unit_price"]

df["hour"] = pd.to_datetime(
    df["transaction_time"], format="%H:%M:%S"
).dt.hour

df["day_name"] = df["transaction_date"].dt.day_name()

df["day_type"] = df["transaction_date"].dt.dayofweek.map(
    lambda x: "Weekend" if x >= 5 else "Weekday"
)

df["month"] = df["transaction_date"].dt.to_period("M").astype(str)

df.head()
```

```python
print("Duplicate complete rows:", df.duplicated().sum())
print("Duplicate transaction IDs:", df["transaction_id"].duplicated().sum())
```

```python
print("Total Revenue:", round(df["revenue"].sum(), 2))
print("Total Transactions:", df["transaction_id"].nunique())
print("Total Items Sold:", df["transaction_qty"].sum())
print("Average Transaction Value:", round(df["revenue"].sum() / df["transaction_id"].nunique(), 2))
```

```python
store_analysis = (
    df.groupby("store_location")
      .agg(
          transactions=("transaction_id", "nunique"),
          items_sold=("transaction_qty", "sum"),
          revenue=("revenue", "sum")
      )
      .sort_values("revenue", ascending=False)
)

store_analysis
```

```python
store_analysis["revenue_percentage"] = (
    store_analysis["revenue"] / df["revenue"].sum() * 100
).round(2)

store_analysis
```

```python
category_analysis = (
    df.groupby("product_category")
      .agg(
          items_sold=("transaction_qty", "sum"),
          revenue=("revenue", "sum")
      )
      .sort_values("revenue", ascending=False)
)

category_analysis["revenue_percentage"] = (
    category_analysis["revenue"] / df["revenue"].sum() * 100
).round(2)

category_analysis
```

```python
top_products = (
    df.groupby(["product_detail", "product_category"])
      .agg(
          items_sold=("transaction_qty", "sum"),
          revenue=("revenue", "sum")
      )
      .sort_values("revenue", ascending=False)
      .head(10)
)

top_products
```

```python
monthly_analysis = (
    df.groupby("month")
      .agg(
          transactions=("transaction_id", "nunique"),
          items_sold=("transaction_qty", "sum"),
          revenue=("revenue", "sum")
      )
      .sort_index()
)

monthly_analysis
```

```python
monthly_analysis["growth_percentage"] = (
    monthly_analysis["revenue"].pct_change() * 100
).round(2)

monthly_analysis
```

```python
monthly_analysis["cumulative_revenue"] = monthly_analysis["revenue"].cumsum()

monthly_analysis
```

```python
day_analysis = (
    df.groupby("day_name")
    .agg(
        transactions=("transaction_id", "nunique"),
        items_sold=("transaction_qty", "sum"),
        revenue=("revenue", "sum")
    )
    .sort_values("revenue", ascending=False)
)

day_analysis
```

```python
hour_analysis = (
    df.groupby("hour")
    .agg(
        transactions=("transaction_id", "nunique"),
        items_sold=("transaction_qty", "sum"),
        revenue=("revenue", "sum")
    )
    .sort_values("revenue", ascending=False)
)

hour_analysis
```

```python
product_type_analysis = (
    df.groupby("product_type")
    .agg(
        transactions=("transaction_id", "nunique"),
        items_sold=("transaction_qty", "sum"),
        revenue=("revenue", "sum")
    )
    .sort_values("revenue", ascending=False)
)

product_type_analysis
```

```python
store_category_analysis = (
    df.groupby(["store_location", "product_category"])
    .agg(
        transactions=("transaction_id", "nunique"),
        items_sold=("transaction_qty", "sum"),
        revenue=("revenue", "sum")
    )
    .sort_values(
        ["store_location", "revenue"],
        ascending=[True, False]
    )
)

store_category_analysis
```

```python
store_product_analysis = (
    df.groupby(["store_location", "product_type"])
    .agg(
        transactions=("transaction_id", "nunique"),
        items_sold=("transaction_qty", "sum"),
        revenue=("revenue", "sum")
    )
    .sort_values(
        ["store_location", "revenue"],
        ascending=[True, False]
    )
)

store_product_analysis
```

```python
category_product_analysis = (
    df.groupby(["product_category", "product_type"])
    .agg(
        transactions=("transaction_id", "nunique"),
        items_sold=("transaction_qty", "sum"),
        revenue=("revenue", "sum")
    )
    .sort_values(
        ["product_category", "revenue"],
        ascending=[True, False]
    )
)

category_product_analysis
```

```python
category_product_analysis["category_revenue_percentage"] = (
    category_product_analysis["revenue"]
    / category_product_analysis.groupby(level=0)["revenue"].transform("sum")
    * 100
).round(2)

category_product_analysis
```

```python
store_category_analysis["store_revenue_percentage"] = (
    store_category_analysis["revenue"]
    / store_category_analysis.groupby(level=0)["revenue"].transform("sum")
    * 100
).round(2)

store_category_analysis
```

```python
hour_daytype_analysis = (
    df.groupby(["day_type", "hour"])
    .agg(
        transactions=("transaction_id", "nunique"),
        items_sold=("transaction_qty", "sum"),
        revenue=("revenue", "sum")
    )
    .sort_values(
        ["day_type", "revenue"],
        ascending=[True, False]
    )
)

hour_daytype_analysis
```

```python
daytype_analysis = (
    df.groupby("day_type")
    .agg(
        transactions=("transaction_id", "nunique"),
        items_sold=("transaction_qty", "sum"),
        revenue=("revenue", "sum")
    )
    .sort_values("revenue", ascending=False)
)

daytype_analysis
```

```python
# ==========================================
# BREWLYTICS - FINAL ANALYSIS SUMMARY
# ==========================================

print("=" * 70)
print("              BREWLYTICS - FINAL ANALYSIS")
print("=" * 70)

# ------------------------------------------
# 1. OVERALL BUSINESS PERFORMANCE
# ------------------------------------------

total_revenue = df["revenue"].sum()
total_transactions = df["transaction_id"].nunique()
total_items = df["transaction_qty"].sum()

print("\n1. OVERALL BUSINESS PERFORMANCE")
print("-" * 70)

print(f"Total Revenue: ${total_revenue:,.2f}")
print(f"Total Transactions: {total_transactions:,}")
print(f"Total Items Sold: {total_items:,}")


# ------------------------------------------
# 2. STORE PERFORMANCE
# ------------------------------------------

best_store = store_analysis["revenue"].idxmax()
best_store_revenue = store_analysis["revenue"].max()

lowest_store = store_analysis["revenue"].idxmin()
lowest_store_revenue = store_analysis["revenue"].min()

print("\n2. STORE PERFORMANCE")
print("-" * 70)

print(f"Best Performing Store: {best_store}")
print(f"Revenue: ${best_store_revenue:,.2f}")

print(f"\nLowest Performing Store: {lowest_store}")
print(f"Revenue: ${lowest_store_revenue:,.2f}")


# ------------------------------------------
# 3. PRODUCT CATEGORY PERFORMANCE
# ------------------------------------------

best_category = category_analysis["revenue"].idxmax()
best_category_revenue = category_analysis["revenue"].max()

lowest_category = category_analysis["revenue"].idxmin()
lowest_category_revenue = category_analysis["revenue"].min()

print("\n3. PRODUCT CATEGORY PERFORMANCE")
print("-" * 70)

print(f"Best Category: {best_category}")
print(f"Revenue: ${best_category_revenue:,.2f}")

print(f"\nLowest Category: {lowest_category}")
print(f"Revenue: ${lowest_category_revenue:,.2f}")


# ------------------------------------------
# 4. PRODUCT TYPE PERFORMANCE
# ------------------------------------------

best_product = product_type_analysis["revenue"].idxmax()
best_product_revenue = product_type_analysis["revenue"].max()

lowest_product = product_type_analysis["revenue"].idxmin()
lowest_product_revenue = product_type_analysis["revenue"].min()

print("\n4. PRODUCT TYPE PERFORMANCE")
print("-" * 70)

print(f"Best Product Type: {best_product}")
print(f"Revenue: ${best_product_revenue:,.2f}")

print(f"\nLowest Product Type: {lowest_product}")
print(f"Revenue: ${lowest_product_revenue:,.2f}")


# ------------------------------------------
# 5. CUSTOMER TRANSACTION METRICS
# ------------------------------------------

average_transaction = total_revenue / total_transactions
average_items = total_items / total_transactions
average_item_revenue = total_revenue / total_items

print("\n5. CUSTOMER TRANSACTION METRICS")
print("-" * 70)

print(f"Average Revenue per Transaction: ${average_transaction:.2f}")
print(f"Average Items per Transaction: {average_items:.2f}")
print(f"Average Revenue per Item: ${average_item_revenue:.2f}")


# ------------------------------------------
# 6. MONTHLY PERFORMANCE
# ------------------------------------------

monthly_revenue = (
    df.groupby(
        df["transaction_date"].dt.to_period("M")
    )["revenue"]
    .sum()
)

best_month = monthly_revenue.idxmax()
best_month_revenue = monthly_revenue.max()

lowest_month = monthly_revenue.idxmin()
lowest_month_revenue = monthly_revenue.min()

print("\n6. MONTHLY PERFORMANCE")
print("-" * 70)

print(f"Best Month: {best_month}")
print(f"Revenue: ${best_month_revenue:,.2f}")

print(f"\nLowest Month: {lowest_month}")
print(f"Revenue: ${lowest_month_revenue:,.2f}")


# ------------------------------------------
# 7. HOURLY PERFORMANCE
# ------------------------------------------

hour_revenue = (
    df.groupby(df["transaction_date"].dt.hour)["revenue"]
    .sum()
)

peak_hour = hour_revenue.idxmax()
peak_hour_revenue = hour_revenue.max()

lowest_hour = hour_revenue.idxmin()
lowest_hour_revenue = hour_revenue.min()

print("\n7. HOURLY PERFORMANCE")
print("-" * 70)

print(f"Peak Revenue Hour: {peak_hour}:00")
print(f"Revenue: ${peak_hour_revenue:,.2f}")

print(f"\nLowest Revenue Hour: {lowest_hour}:00")
print(f"Revenue: ${lowest_hour_revenue:,.2f}")


# ------------------------------------------
# 8. WEEKDAY VS WEEKEND
# ------------------------------------------

weekday_revenue = df.loc[
    df["transaction_date"].dt.dayofweek < 5,
    "revenue"
].sum()

weekend_revenue = df.loc[
    df["transaction_date"].dt.dayofweek >= 5,
    "revenue"
].sum()

print("\n8. WEEKDAY VS WEEKEND")
print("-" * 70)

print(f"Weekday Revenue: ${weekday_revenue:,.2f}")
print(f"Weekend Revenue: ${weekend_revenue:,.2f}")

if weekday_revenue > weekend_revenue:
    print("\nWeekdays generate more revenue than weekends.")
else:
    print(print("\nWeekdays generate significantly more revenue than weekends."))


# ------------------------------------------
# 9. FINAL BUSINESS RECOMMENDATIONS
# ------------------------------------------

print("\n9. BUSINESS RECOMMENDATIONS")
print("-" * 70)

print(f"1. Focus inventory and marketing on the best-performing category: {best_category}.")
print(f"2. Investigate the low performance of {lowest_category}.")
print(f"3. Study the strategies used at {best_store} and apply successful practices to other stores.")
print(f"4. Focus staffing and promotions around the peak hour of {peak_hour}:00.")
print(f"5. Use promotional campaigns to improve performance during weaker periods.")
print(f"6. Encourage combo purchases to increase the average items per transaction.")
print(f"7. Continue monitoring monthly revenue trends for inventory and staffing decisions.")

print("\n" + "=" * 70)
print("                 END OF ANALYSIS")
print("=" * 70)
```

```python
largest_basket = transaction_summary.loc[
    transaction_summary["items_purchased"].idxmax()
]

print("Largest Basket Size:",
      int(largest_basket["items_purchased"]),
      "items")

print("Transaction Value:",
      round(largest_basket["transaction_value"], 2))
```
